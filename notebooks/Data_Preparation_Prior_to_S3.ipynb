```python
import pandas as pd
import glob

# ── 1. Load & combine ──────────────────────────────────────────
files = sorted(glob.glob("Employer Information*.csv"))

dfs = []
for f in files:
    df = pd.read_csv(f, encoding='utf-16', sep='\t', low_memory=False)
    dfs.append(df)
    print(f"Loaded: {f} — {len(df)} rows")

combined = pd.concat(dfs, ignore_index=True)

# ── 2. Clean column names (strip whitespace) ───────────────────
combined.columns = combined.columns.str.strip()

# ── 3. Sort by Fiscal Year ─────────────────────────────────────
combined = combined.sort_values('Fiscal Year').reset_index(drop=True)

# ── 4. Save as UTF-8 CSV ───────────────────────────────────────
combined.to_csv("h1b_combined.csv", index=False, encoding='utf-8')
print("\nSaved: h1b_combined.csv")
```

```python
# ── 5. Sanity Check ────────────────────────────────────────────
print("\n========== SANITY CHECK ==========")

print("\n[ Basic Info ]")
print(f"Total rows      : {len(combined):,}")
print(f"Total columns   : {len(combined.columns)}")
print(f"Column names    : {combined.columns.tolist()}")

print("\n[ Year Range ]")
print(f"Min year : {combined['Fiscal Year'].min()}")
print(f"Max year : {combined['Fiscal Year'].max()}")

print("\n[ Rows per Year ] (sorted)")
print(combined.groupby('Fiscal Year').size().sort_index().to_string())

print("\n[ Missing Values ]")
missing = combined.isnull().sum()
missing_pct = (missing / len(combined) * 100).round(2)
missing_df = pd.DataFrame({'missing_count': missing, 'missing_%': missing_pct})
print(missing_df[missing_df['missing_count'] > 0])

print("\n[ Duplicate Rows ]")
print(f"Duplicate rows: {combined.duplicated().sum():,}")

print("\n[ Sample Data (first 3 rows) ]")
print(combined.head(3).to_string())

print("\n[ Sample Data (last 3 rows) ]")
print(combined.tail(3).to_string())

print("\n========== ALL DONE ==========")
```

```python
import pandas as pd
import glob
import re
import time

COLUMN_MAP = {
    'CASE_NO': 'CASE_NUMBER', 'CASE_NUMBER': 'CASE_NUMBER',
    'CASE_STATUS': 'CASE_STATUS', 'Case_Status': 'CASE_STATUS',
    'DECISION_DATE': 'DECISION_DATE', 'Decision_Date': 'DECISION_DATE',
    'CASE_RECEIVED_DATE': 'RECEIVED_DATE', 'RECEIVED_DATE': 'RECEIVED_DATE',
    'APPLICATION_TYPE': 'APPLICATION_TYPE', 'OCCUPATION_TYPE': 'APPLICATION_TYPE',
    'REFILE': 'REFILE', 'ORIG_FILE_DATE': 'ORIG_FILE_DATE',
    'EMPLOYER_NAME': 'EMPLOYER_NAME', 'Employer_Name': 'EMPLOYER_NAME',
    'EMPLOYER NAME': 'EMPLOYER_NAME', 'EMP_BUSINESS_NAME': 'EMPLOYER_NAME',
    'EMPLOYER_ADDRESS_1': 'EMPLOYER_ADDRESS_1', 'Employer_address_1': 'EMPLOYER_ADDRESS_1',
    'EMPLOYER ADDRESS_1': 'EMPLOYER_ADDRESS_1', 'EMP_ADDR1': 'EMPLOYER_ADDRESS_1',
    'EMPLOYER_CITY': 'EMPLOYER_CITY', 'Employer_City': 'EMPLOYER_CITY',
    'EMPLOYER CITY': 'EMPLOYER_CITY', 'EMP_CITY': 'EMPLOYER_CITY',
    'EMPLOYER_STATE': 'EMPLOYER_STATE', 'Employer_State': 'EMPLOYER_STATE',
    'EMPLOYER STATE': 'EMPLOYER_STATE', 'EMPLOYER_STATE_PROVINCE': 'EMPLOYER_STATE',
    'EMP_STATE': 'EMPLOYER_STATE',
    'EMPLOYER_POSTAL_CODE': 'EMPLOYER_POSTAL_CODE', 'Employer_Postal_Code': 'EMPLOYER_POSTAL_CODE',
    'EMPLOYER POSTAL CODE': 'EMPLOYER_POSTAL_CODE', 'EMP_POSTCODE': 'EMPLOYER_POSTAL_CODE',
    'EMPLOYER_COUNTRY': 'EMPLOYER_COUNTRY', 'EMP_COUNTRY': 'EMPLOYER_COUNTRY',
    'EMPLOYER_NUM_EMPLOYEES': 'EMPLOYER_NUM_EMPLOYEES', 'EMP_NUM_PAYROLL': 'EMPLOYER_NUM_EMPLOYEES',
    'EMPLOYER_YR_ESTAB': 'EMPLOYER_YEAR_ESTABLISHED',
    'EMPLOYER_YEAR_COMMENCED_BUSINESS': 'EMPLOYER_YEAR_ESTABLISHED',
    'EMP_YEAR_COMMENCED_BUSINESS': 'EMPLOYER_YEAR_ESTABLISHED',
    'EMP_YEAR_COMMENCED': 'EMPLOYER_YEAR_ESTABLISHED',
    'EMPLOYER_PHONE': 'EMPLOYER_PHONE', 'EMP_PHONE': 'EMPLOYER_PHONE',
    '2007_NAICS_US_CODE': 'NAICS_CODE', '2007 NAICS US CODE': 'NAICS_CODE',
    '2007_NAICS_US_Code': 'NAICS_CODE', 'NAICS_US_CODE': 'NAICS_CODE',
    'NAICS_CODE': 'NAICS_CODE', 'EMP_NAICS': 'NAICS_CODE',
    '2007_NAICS_US_TITLE': 'NAICS_TITLE', '2007 NAICS US TITLE': 'NAICS_TITLE',
    'NAICS_US_TITLE': 'NAICS_TITLE', 'US_ECONOMIC_SECTOR': 'US_ECONOMIC_SECTOR',
    'JOB_INFO_JOB_TITLE': 'JOB_TITLE', 'JOB_TITLE': 'JOB_TITLE',
    'PW_JOB_TITLE_9089': 'JOB_TITLE', 'PW_Job_Title_9089': 'JOB_TITLE',
    'PW JOB TITLE 9089': 'JOB_TITLE', 'PW_JOB_TITLE': 'JOB_TITLE',
    'JOB_INFO_WORK_CITY': 'WORKSITE_CITY', 'WORKSITE_CITY': 'WORKSITE_CITY',
    'PRIMARY_WORKSITE_CITY': 'WORKSITE_CITY',
    'JOB_INFO_WORK_STATE': 'WORKSITE_STATE', 'WORKSITE_STATE': 'WORKSITE_STATE',
    'PRIMARY_WORKSITE_STATE': 'WORKSITE_STATE',
    'JOB_INFO_WORK_POSTAL_CODE': 'WORKSITE_POSTAL_CODE',
    'WORKSITE_POSTAL_CODE': 'WORKSITE_POSTAL_CODE',
    'PRIMARY_WORKSITE_POSTAL_CODE': 'WORKSITE_POSTAL_CODE',
    'JOB_INFO_EDUCATION': 'MIN_EDUCATION', 'MINIMUM_EDUCATION': 'MIN_EDUCATION',
    'JOB_INFO_MAJOR': 'MAJOR_FIELD_OF_STUDY', 'MAJOR_FIELD_OF_STUDY': 'MAJOR_FIELD_OF_STUDY',
    'WAGE_OFFER_FROM_9089': 'WAGE_OFFER_FROM', 'WAGE_OFFERED_FROM_9089': 'WAGE_OFFER_FROM',
    'WAGE_OFFER_FROM': 'WAGE_OFFER_FROM', 'JOB_OPP_WAGE_FROM': 'WAGE_OFFER_FROM',
    'WAGE_OFFER_TO_9089': 'WAGE_OFFER_TO', 'WAGE_OFFERED_TO_9089': 'WAGE_OFFER_TO',
    'WAGE_OFFER_TO': 'WAGE_OFFER_TO', 'JOB_OPP_WAGE_TO': 'WAGE_OFFER_TO',
    'WAGE_OFFER_UNIT_OF_PAY_9089': 'WAGE_UNIT_OF_PAY',
    'WAGE_OFFERED_UNIT_OF_PAY_9089': 'WAGE_UNIT_OF_PAY',
    'WAGE_OFFER_UNIT_OF_PAY': 'WAGE_UNIT_OF_PAY', 'JOB_OPP_WAGE_PER': 'WAGE_UNIT_OF_PAY',
    'PW_AMOUNT_9089': 'PW_WAGE', 'PW_WAGE': 'PW_WAGE',
    'PW_UNIT_OF_PAY_9089': 'PW_UNIT_OF_PAY', 'PW_UNIT_OF_PAY': 'PW_UNIT_OF_PAY',
    'PW_SOC_CODE': 'SOC_CODE', 'PWD_SOC_CODE': 'SOC_CODE', 'PW SOC CODE': 'SOC_CODE',
    'PW_SOC_TITLE': 'SOC_TITLE', 'PWD_SOC_TITLE': 'SOC_TITLE',
    'PW_LEVEL_9089': 'PW_SKILL_LEVEL', 'PW_SKILL_LEVEL': 'PW_SKILL_LEVEL',
    'COUNTRY_OF_CITZENSHIP': 'COUNTRY_OF_CITIZENSHIP',
    'COUNTRY_OF_CITIZENSHIP': 'COUNTRY_OF_CITIZENSHIP',
    'Country_of_Citizenship': 'COUNTRY_OF_CITIZENSHIP',
    'COUNTRY OF CITZENSHIP': 'COUNTRY_OF_CITIZENSHIP',
    'CLASS_OF_ADMISSION': 'CLASS_OF_ADMISSION', 'Class_of_Admission': 'CLASS_OF_ADMISSION',
    'CLASS OF ADMISSION': 'CLASS_OF_ADMISSION',
    'FW_INFO_BIRTH_COUNTRY': 'FW_BIRTH_COUNTRY',
    'FOREIGN_WORKER_BIRTH_COUNTRY': 'FW_BIRTH_COUNTRY',
    'FOREIGN_WORKER_INFO_CITY': 'FW_CITY', 'FOREIGN_WORKER_INFO_STATE': 'FW_STATE',
    'FOREIGN_WORKER_INFO_EDUCATION': 'FW_EDUCATION', 'FOREIGN_WORKER_EDUCATION': 'FW_EDUCATION',
    'FOREIGN_WORKER_INFO_MAJOR': 'FW_MAJOR',
    'FW_INFO_REQ_EXPERIENCE': 'FW_REQUIRED_EXPERIENCE',
    'FOREIGN_WORKER_REQ_EXPERIENCE': 'FW_REQUIRED_EXPERIENCE',
}

def extract_year(filename):
    match = re.search(r'FY(\d{2,4})', filename, re.IGNORECASE)
    if match:
        yr = match.group(1)
        return int('20' + yr) if len(yr) == 2 else int(yr)
    return None

files = sorted(glob.glob("*.xlsx"))
total = len(files)
print(f"Total files: {total}\n")

dfs = []
start_total = time.time()

for i, f in enumerate(files, 1):
    start = time.time()
    print(f"[{i}/{total}] Reading: {f} ...", end=' ', flush=True)
    try:
        # read_excel with calamine engine — much faster than openpyxl
        try:
            df = pd.read_excel(f, engine='calamine')
        except:
            df = pd.read_excel(f, engine='openpyxl')

        df = df.loc[:, ~df.columns.duplicated()]
        df = df.rename(columns=COLUMN_MAP)
        df = df.loc[:, ~df.columns.duplicated()]
        df['SOURCE_FILE'] = f
        df['FISCAL_YEAR'] = extract_year(f)
        dfs.append(df)
        elapsed = time.time() - start
        print(f"done ({len(df):,} rows, {elapsed:.1f}s)")
    except Exception as e:
        print(f"ERROR: {e}")

print(f"\nCombining...")
combined = pd.concat(dfs, ignore_index=True, sort=False)
combined = combined.sort_values('FISCAL_YEAR').reset_index(drop=True)

print(f"Saving...")
combined.to_csv("perm_combined.csv", index=False, encoding='utf-8')

total_time = time.time() - start_total
print(f"\nDone! Total time: {total_time:.1f}s")
print(f"Total rows    : {len(combined):,}")
print(f"Total columns : {len(combined.columns)}")
print(f"Year range    : {combined['FISCAL_YEAR'].min()} ~ {combined['FISCAL_YEAR'].max()}")
print(f"--> Ready to upload to S3: visa-raw-data-group1-apan/perm/")
```

```python
import pandas as pd
import os

# ================================================================
# PERM Combined Data — Remove Unmapped Columns & Reduce File Size
# APAN 5450: Cloud Computing | Group 1
# Description: Keeps only standardized columns from the combined
#              PERM dataset. Removes unmapped/duplicate columns
#              that inflated the file to 2.3GB.
# ================================================================

print("## Step 1: Loading perm_combined.csv...\n")
df = pd.read_csv("perm_combined.csv", low_memory=False)
print(f"  Before — Rows: {len(df):,} | Columns: {len(df.columns)}")

# ── Step 2: Define standard columns to keep ───────────────────
standard_cols = [
    # Case info
    'CASE_NUMBER', 'CASE_STATUS', 'DECISION_DATE', 'RECEIVED_DATE',
    'APPLICATION_TYPE', 'REFILE', 'ORIG_FILE_DATE',
    # Employer info
    'EMPLOYER_NAME', 'EMPLOYER_ADDRESS_1', 'EMPLOYER_CITY',
    'EMPLOYER_STATE', 'EMPLOYER_POSTAL_CODE', 'EMPLOYER_COUNTRY',
    'EMPLOYER_PHONE', 'EMPLOYER_NUM_EMPLOYEES', 'EMPLOYER_YEAR_ESTABLISHED',
    # Industry
    'NAICS_CODE', 'NAICS_TITLE', 'US_ECONOMIC_SECTOR',
    # Job info
    'JOB_TITLE', 'WORKSITE_CITY', 'WORKSITE_STATE', 'WORKSITE_POSTAL_CODE',
    'MIN_EDUCATION', 'MAJOR_FIELD_OF_STUDY',
    # Wage info
    'WAGE_OFFER_FROM', 'WAGE_OFFER_TO', 'WAGE_UNIT_OF_PAY',
    'PW_WAGE', 'PW_UNIT_OF_PAY', 'PW_SKILL_LEVEL',
    # Occupation
    'SOC_CODE', 'SOC_TITLE',
    # Foreign worker info
    'COUNTRY_OF_CITIZENSHIP', 'CLASS_OF_ADMISSION',
    'FW_BIRTH_COUNTRY', 'FW_CITY', 'FW_STATE',
    'FW_EDUCATION', 'FW_MAJOR', 'FW_REQUIRED_EXPERIENCE',
    # Metadata
    'SOURCE_FILE', 'FISCAL_YEAR'
]

# ── Step 3: Keep only available standard columns ───────────────
print("\n## Step 2: Filtering to standard columns only...\n")
available = [c for c in standard_cols if c in df.columns]
missing_std = [c for c in standard_cols if c not in df.columns]

print(f"  Standard cols defined  : {len(standard_cols)}")
print(f"  Found in data          : {len(available)}")
if missing_std:
    print(f"  Not found (will skip)  : {missing_std}")

df_clean = df[available].copy()
print(f"\n  After — Rows: {len(df_clean):,} | Columns: {len(df_clean.columns)}")

# ── Step 4: Save cleaned version ──────────────────────────────
print("\n## Step 3: Saving cleaned CSV...\n")
df_clean.to_csv("perm_combined.csv", index=False, encoding='utf-8')
size_mb = os.path.getsize("perm_combined.csv") / 1024 / 1024
print(f"  Saved: perm_combined.csv")
print(f"  File size: {size_mb:.1f} MB")

# ── Step 5: Sanity Check ──────────────────────────────────────
print("\n## Step 4: Sanity Check\n")

print("### Basic Info")
print(f"  Total rows    : {len(df_clean):,}")
print(f"  Total columns : {len(df_clean.columns)}")
print(f"  Columns       : {df_clean.columns.tolist()}")

print("\n### Year Coverage")
print(f"  Min year : {df_clean['FISCAL_YEAR'].min()}")
print(f"  Max year : {df_clean['FISCAL_YEAR'].max()}")

print("\n### Rows per Fiscal Year")
print(df_clean.groupby('FISCAL_YEAR').size().sort_index().to_string())

print("\n### Missing Values")
missing = df_clean.isnull().sum()
missing_pct = (missing / len(df_clean) * 100).round(2)
print(pd.DataFrame({'missing_count': missing, 'missing_%': missing_pct}).to_string())

print("\n### Duplicate Rows")
print(f"  Duplicates: {df_clean.duplicated().sum():,}")

print("\n### Case Status Distribution")
if 'CASE_STATUS' in df_clean.columns:
    print(df_clean['CASE_STATUS'].value_counts().head(10).to_string())

print("\n### Sample (first 3 rows)")
print(df_clean.head(3).to_string())

print("\n## Summary")
print(f"  File saved    : perm_combined.csv")
print(f"  File size     : {size_mb:.1f} MB")
print(f"  Total rows    : {len(df_clean):,}")
print(f"  Total columns : {len(df_clean.columns)}")
print(f"  Year range    : {df_clean['FISCAL_YEAR'].min()} ~ {df_clean['FISCAL_YEAR'].max()}")
print(f"  --> Ready to upload to S3: visa-raw-data-group1-apan/perm/")
```
